```python
import pandas as pd
pd.options.display.max_rows = 10

unames = ['user_id', 'gender', 'age', 'occupation', 'zip']
users = pd.read_table(r'C:\Users\ipand\Documents\data_science\users.dat', sep ='::', header =None , names= unames)

rnames = ['user_id', 'movie_id', 'rating', 'timestamp']
ratings = pd.read_table(r'C:\Users\ipand\Documents\data_science\ratings.dat', sep='::',header=None, names=rnames)
mnames = ['movie_id', 'title', 'genres']
movies = pd.read_table(r'C:\Users\ipand\Documents\data_science\movies.dat', sep='::',
header=None, names=mnames)
```

### Merging tables 

```python
merged_data = pd.merge(pd.merge(ratings,users),movies) #The pd.merge outside merges the ratings and users dataframes, and then the result is merged with the movies dataframe. This creates a single dataframe that contains all the information from the three original dataframes.
merged_data
```

```python
merged_data.iloc[0]  #The iloc method is used to access a specific row in the merged_data dataframe. In this case, it retrieves the first row (index 0) of the dataframe, which contains information about a specific rating, user, and movie.
```

```python
mean_ratings = merged_data.pivot_table('rating', index ='title', columns='gender',aggfunc='mean')  #The pivot_table method is used to create a new dataframe called mean_ratings. It calculates the mean rating for each movie title, grouped by gender. The 'rating' column is used as the values to be aggregated, 'title' is set as the index, and 'gender' is set as the columns. The aggfunc parameter specifies that the mean function should be used for aggregation. 
mean_ratings
```

```python
# In order to get titles with ratings more than 250, we can use the groupby method to group the merged_data dataframe by movie title and then use the size method to count the number of ratings for each title. We can then filter the mean_ratings dataframe to only include titles with more than 250 ratings.

ratings_by_title = merged_data.groupby('title').size()
filtered_ratings = ratings_by_title[ratings_by_title > 250]
mean_ratings = mean_ratings.loc[filtered_ratings.index]
```

```python
#To get the top  10 female and male movies, we can sort the F and M columns in descending order and select the top 10 rows using the head method. This will give us the top 10 movies with the highest average ratings for each gender.
top_female_movies = mean_ratings.sort_values(by= 'F', ascending=False).head(10)
top_male_movies= mean_ratings.sort_values(by ='M', ascending =False).head(10)
```

```python
top_female_movies, top_male_movies
```

```python
mean_ratings['diff'] = mean_ratings['M'] - mean_ratings['F']
mean_ratings['diff_absolute'] = (mean_ratings['M'] - mean_ratings['F']).abs()

mean_ratings[['F','M','diff']].sort_values(by = 'diff', ascending =False).head(10)

#To get the top 10 movies with the largest difference in ratings between males and females and the top 10 movies with the smallest difference in ratings between males and females
```
